```python
languages = ["Java", "Python", "C++", "Go", "JavaScript"]

#first language, last language, first 3 languages, last 2 languages, length of the list

print(languages[0])
print(languages[-1])
print(languages[0:3])
print(languages[-2:])
print(len(languages))
```

```python
square = []
numbers = [1, 2, 3, 4, 5]

for number in numbers:
  square.append( number * number)

print(square)

square = [ number*number for number in numbers ]
print(square)
```

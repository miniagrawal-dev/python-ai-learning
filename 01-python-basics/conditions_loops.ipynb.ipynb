```python
documents = [
    {"title": "Python", "score": 0.72},
    {"title": "RAG", "score": 0.95},
    {"title": "Agents", "score": 0.88},
    {"title": "LLMs", "score": 0.93}
]

for document in documents :
  if(document["score"] >= 0.90):
    print(document["score"])

for document in documents :
  if(document["score"] >= 0.90):
    print(document["title"])
    break

for document in documents :
  if(document["score"] < 0.80):
    continue;
  else :
    print(document["score"])
```

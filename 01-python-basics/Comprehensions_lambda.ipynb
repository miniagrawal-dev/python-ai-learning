```python
def say_hello(name) :
    print("hello", name)

say_hello("mini")

def add(a, b):
   return a+b

print(add(1,2))

def getUser(user, age):
  return user, age

user, age = getUser("mini", 30)

print(user, age)
```

```python
numbers = [1, 2, 3, 4, 5]

squares_of_even = [
    number*number
    for number in numbers
    if number %2 == 0
]

print(squares_of_even)

squares = {
    number: number*number
    for number in numbers
}

print(squares)
```

```python
matrix = [
    [1, 2],
    [3, 4]
]

flattened = [
    number
    for row in matrix
    for number in row
]

print(flattened)
```

```python
def square(x):
    return x * x

#lambda
#lambda arguments: expression
square = lambda x: x * x

print(square(5))
```

```python
documents = [
    {"title": "Python", "score": 0.72},
    {"title": "RAG", "score": 0.95},
    {"title": "Agents", "score": 0.88}
]

sorted_documents = sorted(
    documents,
    key=lambda document: document["score"]
)

print(documents)
print(sorted_documents)
```

```python
numbers = [1, 2, 3, 4]
def square(x):
    return x * x

result = list(map(square, numbers))
print(result)
```
